```python
df.columns=df.columns.str.title()
```

 **Feature Engineering**

Create Age Groups

```python
df['AgeGroup']=pd.cut(df['Age'],
bins=[18,25,35,45,55,60],
labels=['18-25','26-35','36-45','46-55','56-60'])
```

Experience Level

```python
df.columns
```

```python
df['ExperienceLevel']=pd.cut(df['Totalworkingyears'],
bins=[0,5,10,20,40],
labels=['Junior','Mid','Senior','Expert'])
```

Income Category

```python
df['IncomeCategory']=pd.qcut(df['Monthlyincome'],4,
labels=['Low','Medium','High','Very High'])
```
